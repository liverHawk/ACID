```python
import pandas as pd
from glob import glob
```

```python
path = './dataset/CICIDS2017_improved'
files = glob(path + '/*.csv')

dfs = []
for file in files:
    df = pd.read_csv(file)
    dfs.append(df)

df = pd.concat(dfs)
```

```python
df["Protocol_0"] = (df["Protocol"] == 0).astype(int)
df["Protocol_6"] = (df["Protocol"] == 6).astype(int)
df["Protocol_17"] = (df["Protocol"] == 17).astype(int)

df = df.drop(columns=["Protocol", "Attempted Category", "ICMP Type", "ICMP Code", "Flow ID", "Timestamp", "Src IP", "Src Port", "Dst IP", "Dst Port", "id"])
```

```python
labels = df["Label"].unique()
for label in labels:
    if "Attempted" in label:
        df.loc[df["Label"] == label, "Label"] = "BENIGN"
    if "Web Attack" in label:
        df.loc[df["Label"] == label, "Label"] = "Web Attack"
    if "Portscan" in label:
        df.loc[df["Label"] == label, "Label"] = "Portscan"
    if "Infiltration" in label:
        df.loc[df["Label"] == label, "Label"] = "Infiltration"
    if "DoS" in label and label != "DDoS":
        df.loc[df["Label"] == label, "Label"] = "DoS"
```

```python
import os
output_path = './dataset/processed'
os.makedirs(output_path, exist_ok=True)

# save dataset (split 100000raw)
for idx, i in enumerate(range(0, len(df), 100000)):
    save_path = os.path.join(output_path, f'{idx}.csv')
    df.iloc[i:i+100000].to_csv(save_path, index=False)

print(f"Done saving dataset to {output_path}")
```
